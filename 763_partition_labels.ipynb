```python
from typing import List
```

Идея:
- Вначале для каждого уникального символа строки считаем индексы первого и последнего вхождения, таким образом у нас получаются интервалы
- Используем подход из задачи 56 merge intervals и мерджим их в непересекающиеся. Затем считаем их длины. Это и будет соответствовать длинам искомых подпоследовательностей

```python
def partitionLabels(s: str) -> List[int]:
    char_to_interval = {}
    unique_characters_order = []
    for i in range(len(s)):
        if s[i] not in char_to_interval:
            char_to_interval[s[i]] = [i, i]
            unique_characters_order.append(s[i])
        else:
            char_to_interval[s[i]][1] = i

    merged_intervals = []
    result=[]
    for c in unique_characters_order:
        interval = char_to_interval[c]
        if merged_intervals and interval[0] <= merged_intervals[-1][1]:
            merged_intervals[-1][1] = max(merged_intervals[-1][1], interval[1])
            result[-1] = merged_intervals[-1][1] - merged_intervals[-1][0] +1
        else:
            merged_intervals.append(interval)  
            result.append(merged_intervals[-1][1] - merged_intervals[-1][0] +1)
    

    return result
```

```python
s = "ababcbacadefegdehijhklij"
print(partitionLabels(s)) #[9,7,8]
```

```python
s = "eccbbbbdec"
print(partitionLabels(s)) #[10]
```
