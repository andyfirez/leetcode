```python
from typing import List
```

```python
def searchInsert(nums: List[int], target: int) -> int:
    left = 0
    right=len(nums)-1

    while left <=right:
        mid = (right+left)//2

        if target == nums[mid]:
            return mid
        elif (target < nums[mid] and (mid == 0 or target > nums[mid - 1])):
            return mid
        elif target > nums[mid] and mid == len(nums) - 1:
            return mid + 1    
        elif target > nums[mid]:
            left = mid+1
        else:
            right = mid-1
```

```python
nums = [1,3,5,6]
target = 5
print(searchInsert(nums, target))
```
