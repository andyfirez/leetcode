```python
from typing import Optional
```

```python
class TreeNode:
    def __init__(self, val=0, left=None, right=None):
        self.val = val
        self.left = left
        self.right = right
```

```python
def list_to_tree(tree_list: list) -> Optional[TreeNode]:
    if not tree_list:
        return None

    def traverse(pointer: int) -> Optional[TreeNode]:
       if pointer > len(tree_list)-1:
          return None
       
       left_child = traverse(2*pointer + 1)
       right_child = traverse(2*pointer + 2)
       return TreeNode(tree_list[pointer], left_child, right_child)

    return traverse(0)
```

```python
def maxDepth(root: Optional[TreeNode]) -> int:
    if not root:
        return 0

    def traverse(node: TreeNode) -> int:
        if not node:
            return 0

        depth_left = traverse(node.left) + 1
        depth_right = traverse(node.right) + 1

        return max(depth_left, depth_right)

    return traverse(root)
    
```

```python
tree = list_to_tree([3,9,20,None,None,15,7])
print(maxDepth(tree))
```

```python
tree = list_to_tree([1,None,2])
print(maxDepth(tree))
```
