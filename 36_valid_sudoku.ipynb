```python
from typing import List
```

```python
def isValidSudoku(board: List[List[str]]) -> bool:
    row_sets ={}
    column_sets={}
    subsquare_sets={}

    for i in range(9):
        for j in range(9):
            if i not in row_sets:
                row_sets[i] = set()

            if j not in column_sets:
                column_sets[j] = set()

            subsquare_index = 3*(i//3) + j//3
            if subsquare_index not in subsquare_sets:
                subsquare_sets[subsquare_index] = set()

            if board[i][j] == '.':
                continue

            if board[i][j] in row_sets[i] or board[i][j] in column_sets[j] or board[i][j] in subsquare_sets[subsquare_index]:
                return False

            row_sets[i].add(board[i][j])
            column_sets[j].add(board[i][j])
            subsquare_sets[subsquare_index].add(board[i][j])

    return True
```

```python
board = [["5","3",".",".","7",".",".",".","."]
,["6",".",".","1","9","5",".",".","."]
,[".","9","8",".",".",".",".","6","."]
,["8",".",".",".","6",".",".",".","3"]
,["4",".",".","8",".","3",".",".","1"]
,["7",".",".",".","2",".",".",".","6"]
,[".","6",".",".",".",".","2","8","."]
,[".",".",".","4","1","9",".",".","5"]
,[".",".",".",".","8",".",".","7","9"]]

print(isValidSudoku(board))
```
