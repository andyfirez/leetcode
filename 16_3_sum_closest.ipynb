Идея:
- аналогично предыдущей задаче
- считаем разницу между таргетом и нашей суммой и сохраняем минимум
- двигаем указатели по тому же принципу

```python
def threeSumClosest(nums: list[int], target: int) -> int:
    nums.sort()
    n = len(nums)
    sum_3 = nums[0] + nums[1] + nums[2]
    min_difference = target - sum_3
    for i in range(n-2):
        # if i > 0 and nums[i] == nums[i - 1]:
        #     continue
        left_pointer = i+1
        right_pointer = n-1
        target_sum = target-nums[i]
        while left_pointer < right_pointer:
            difference = target_sum - nums[left_pointer] - nums[right_pointer]
            # min_difference = min(min_difference, abs(difference))
            if abs(difference) < min_difference:
                min_difference = abs(difference)
                sum_3 = nums[i] + nums[left_pointer] + nums[right_pointer]
            if difference == 0:
                return target
            elif difference > 0:
                left_pointer+=1
            else:
                right_pointer-=1

    return sum_3
```

```python
nums = [-1,2,1,-4]
target = 1
print(threeSumClosest(nums, target))
```
