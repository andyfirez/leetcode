```python
from typing import Optional, List
```

```python
class TreeNode:
    def __init__(self, val=0, left=None, right=None):
        self.val = val
        self.left = left
        self.right = right
```

```python
def list_to_tree(tree_list: list) -> Optional[TreeNode]:
    if not tree_list:
        return None

    def traverse(pointer: int) -> Optional[TreeNode]:
       if pointer > len(tree_list)-1:
          return None
       
       left_child = traverse(2*pointer + 1)
       right_child = traverse(2*pointer + 2)
       if not tree_list[pointer]:
           return None 
       return TreeNode(tree_list[pointer], left_child, right_child)

    return traverse(0)
```

```python
def rightSideView(root: Optional[TreeNode]) -> List[int]:
    if not root:
        return []
    
    right_nodes = []

    def traverse(node: TreeNode, level: int):
        if not node:
            return
        
        if level == len(right_nodes):
            right_nodes.append(None)

        right_nodes[level] = node.val
        traverse(node.left, level+1)
        traverse(node.right, level+1)

    traverse(root, 0)
    return right_nodes
```

```python
tree_array = [1,2,3,None,5,None,4]
print(rightSideView(list_to_tree(tree_array)))
```

```python
tree_array = [1,2,3,4,None,None,None,5]
print(rightSideView(list_to_tree(tree_array)))
```
