```python
def isPalindrome(x: int) -> bool:
    string = str(x)
    str_reverse = string[::-1]

    return string == str_reverse

isPalindrome(10)
```

```python
def isPalindrome(x: int) -> bool:
    if x < 0:
        return False
    number = x
    reversed_number = 0
    while number > 0:
        reversed_number = reversed_number * 10 + number % 10
        number = number // 10
    return x == reversed_number

isPalindrome(-121)
```
