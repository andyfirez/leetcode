```python
from typing import List
```

Решение 1:
- Ищем значение k используя бинарный поиск
- Используем значение k для преобразования индекса исходного массива в индекс сдвинутого, чтобы определить индекс элемента, который в изначальном массиве был в середине

```python
def get_rotated_index(old_index: int, k: int, n: int) -> int:
    return (old_index + k) % n
    
    
def search_k(nums: List[int]) -> int:
    left = 0
    right = len(nums) - 1

    if nums[left] <= nums[right]:
        return 0
    
    while left <= right:
        mid = (left + right) // 2

        # Проверяем, нашли ли точку сдвига
        if mid > 0 and nums[mid] < nums[mid - 1]:
            return mid
        if mid < len(nums) - 1 and nums[mid] > nums[mid + 1]:
            return mid + 1
        
        # Определяем, в какой половине искать
        if nums[mid] >= nums[left]:
            left = mid + 1
        else:
            right = mid - 1



def search(nums: List[int], target: int) -> int:
    if not nums:
        return -1
        
    # Шаг 1: Находим индекс начала (k)
    k = search_k(nums)
    
    # Шаг 2: Бинарный поиск с преобразованием индексов
    left, right = 0, len(nums) - 1
    
    while left <= right:
        mid = (left + right) // 2
        
        # Преобразуем mid в реальный индекс в сдвинутом массиве
        real_mid = get_rotated_index(mid, k, len(nums))
        
        if nums[real_mid] == target:
            return real_mid
        
        if nums[real_mid] < target:
            left = mid + 1
        else:
            right = mid - 1
            
    return -1
```

```python
nums = [4,5,6,7,0,1,2]
target = 0
print(search(nums, target))
```

Решение 2 (более компактное):
- Какой бы индекс сдвинутого массива мы бы не взяли, у нас всегда либо левая либо правая часть будет отсортированной, а другая - нет
- В классическом бинарном поиске мы сравнивали таргет со средним значением и по результатам переходили в соотв. половину массива. Здесь это не работает, поскольку весь массив не отсортирован. Однако, мы можем сделать проверку по отсортированной части
- таким образом, алгоритм сначала определяет, какая половина отсортирована, и потом проверяем, находится ли таргет в ней. И совершаем сдвиг границ как обычно

```python
def search(self, nums, target):
    l, r = 0, len(nums) - 1
    while l <= r:
        mid = l + (r - l) // 2
        if target == nums[mid]:
            return mid
        
        # Если числа между mid и r отсортированы
        if nums[mid] < nums[r]:
            # Если таргет находится в отсортированной части
            if nums[mid] < target <= nums[r]:
                l = mid + 1
            else:
                r = mid - 1
        # Если числа между l и mid отсортированы
        else:
            # Если таргет находится в отсортированной части
            if nums[l] <= target < nums[mid]:
                r = mid - 1
            else:
                l = mid + 1
    return -1
```
