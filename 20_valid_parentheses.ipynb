```python
from collections import deque
```

```python
def isValid(s: str) -> bool:
    stack = deque()
    for c in s:
        if c == '(' or c == '[' or c == '{':
            stack.append(c)
        else:
            try:
                last_bracket = stack.pop()
            except:
                return False
            if (c == ')' and last_bracket != '(') or (c == ']' and last_bracket != '[') or (c == '}' and last_bracket != '{'):
                return False
    return len(stack) == 0
```

```python
s = "()"
print(isValid(s))
```

```python
s = "()[]{}"
print(isValid(s))
```

```python
s = "(]"
print(isValid(s))
```

```python
s = "([])"
print(isValid(s))
```

```python
s = "([)]"
print(isValid(s))
```
