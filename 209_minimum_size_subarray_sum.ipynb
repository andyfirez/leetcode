```python
from typing import List
```

```python
def minSubArrayLen(target: int, nums: List[int]) -> int:
    p1=0
    p2=0
    s = nums[p2]
    flag = False
    min_length = len(nums)

    while p2 < len(nums):
        if s >= target:
            min_length = min(min_length, p2-p1+1)
            flag = True
            if p1 < p2:
                s-=nums[p1]
                p1+=1
            else:
                p2+=1
                if p2 < len(nums):
                    s+=nums[p2]
        else:
            p2+=1
            if p2 < len(nums):
                s+=nums[p2]


    if flag:
        return min_length
    else:
        return 0
```

```python
target = 7
nums = [2,3,1,2,4,3]
print(minSubArrayLen(target, nums))
```
