```python
from collections import deque


# Граф задаём списком смежности: вершина -> список соседей
# Пример ниже — неориентированный граф (рёбра двусторонние)
graph: dict[int, list[int]] = {
    0: [1, 2],
    1: [0, 3, 4],
    2: [0, 5],
    3: [1],
    4: [1, 5],
    5: [2, 4],
}
```

```python
def bfs_recursive(graph: dict[int, list[int]], start: int) -> list[int]:
    # BFS (обход в ширину) — рекурсивная версия
    # Идея: обрабатываем уровень целиком, следующий уровень передаём в рекурсию
    # В графе обязателен visited: иначе цикл (A → B → A) зациклит обход
    if start not in graph:
        return []

    result: list[int] = []
    visited: set[int] = {start}

    def traverse(level: list[int]) -> None:
        if not level:
            return

        next_level: list[int] = []
        for node in level:
            result.append(node)
            for neighbor in graph.get(node, []):
                if neighbor not in visited:
                    visited.add(neighbor)
                    next_level.append(neighbor)

        traverse(next_level)

    traverse([start])
    return result
```

```python
def bfs_iterative(graph: dict[int, list[int]], start: int) -> list[int]:
    # BFS (обход в ширину) — итеративная версия
    # Идея: очередь (FIFO) — сначала все соседи текущей вершины, потом их соседи
    if start not in graph:
        return []

    result: list[int] = []
    visited: set[int] = {start}
    queue = deque([start])

    while queue:
        node = queue.popleft()  # берём вершину из начала очереди
        result.append(node)
        for neighbor in graph.get(node, []):
            if neighbor not in visited:
                visited.add(neighbor)  # помечаем сразу, чтобы не положить в очередь дважды
                queue.append(neighbor)

    return result
```

```python
def dfs_recursive(graph: dict[int, list[int]], start: int) -> list[int]:
    # DFS (обход в глубину) — рекурсивная версия
    # Идея: уходим по ребру до конца, затем возвращаемся (call stack = стек обхода)
    if start not in graph:
        return []

    result: list[int] = []
    visited: set[int] = set()

    def traverse(node: int) -> None:
        if node in visited:
            return

        visited.add(node)
        result.append(node)
        for neighbor in graph.get(node, []):
            traverse(neighbor)

    traverse(start)
    return result
```

```python
def dfs_iterative(graph: dict[int, list[int]], start: int) -> list[int]:
    # DFS (обход в глубину) — итеративная версия
    # Идея: явный стек (LIFO) вместо стека вызовов рекурсии
    if start not in graph:
        return []

    result: list[int] = []
    visited: set[int] = set()
    stack = [start]

    while stack:
        node = stack.pop()  # берём вершину с вершины стека
        if node in visited:
            continue

        visited.add(node)
        result.append(node)
        # Соседей кладём в обратном порядке — при pop() первый сосед обработается первым
        for neighbor in reversed(graph.get(node, [])):
            if neighbor not in visited:
                stack.append(neighbor)

    return result
```

```python
print("BFS recursive:", bfs_recursive(graph, 0))
print("BFS iterative:", bfs_iterative(graph, 0))
print("DFS recursive:", dfs_recursive(graph, 0))
print("DFS iterative:", dfs_iterative(graph, 0))
```
