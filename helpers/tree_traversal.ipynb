```python
from collections import deque


class TreeNode:
    def __init__(self, val=0, left=None, right=None):
        self.val = val
        self.left = left
        self.right = right
```

```python
def bfs_recursive(root: TreeNode) -> list[list[int]]:
    # BFS (обход в ширину) — рекурсивная версия
    # Идея: передаём номер уровня и складываем узлы в result[level]
    if not root:
        return []

    result: list[list[int]] = []

    def traverse(node: TreeNode, level: int) -> None:
        if not node:
            return

        # Если текущий уровень ещё не создан — добавляем новый список
        if level == len(result):
            result.append([])

        # Записываем значение узла на его уровень
        result[level].append(node.val)
        # Рекурсивно обходим детей на следующем уровне
        traverse(node.left, level + 1)
        traverse(node.right, level + 1)

    traverse(root, 0)
    return result
```

```python
def bfs_iterative(root: TreeNode) -> list[list[int]]:
    # BFS (обход в ширину) — итеративная версия
    # Идея: очередь (FIFO) — сначала обрабатываем все узлы текущего уровня
    if not root:
        return []

    result = []
    queue = deque([root])  # очередь узлов для обхода

    while queue:
        level = []
        # len(queue) на этом шаге = количество узлов на текущем уровне
        for _ in range(len(queue)):
            node = queue.popleft()  # берём узел из начала очереди
            level.append(node.val)
            # Добавляем детей в конец очереди — они будут обработаны на след. уровне
            if node.left:
                queue.append(node.left)
            if node.right:
                queue.append(node.right)
        result.append(level)

    return result
```

```python
def dfs_recursive(root: TreeNode, order: str = "preorder") -> list[int]:
    # DFS (обход в глубину) — рекурсивная версия
    # Идея: уходим в поддерево до конца, затем возвращаемся (call stack = стек обхода)
    if not root:
        return []

    result = []

    def traverse(node: TreeNode) -> None:
        if not node:
            return

        if order == "preorder":
            # Корень → левое → правое
            result.append(node.val)
            traverse(node.left)
            traverse(node.right)
        elif order == "inorder":
            # Левое → корень → правое
            traverse(node.left)
            result.append(node.val)
            traverse(node.right)
        elif order == "postorder":
            # Левое → правое → корень
            traverse(node.left)
            traverse(node.right)
            result.append(node.val)
        else:
            raise ValueError("order must be 'preorder', 'inorder', or 'postorder'")

    traverse(root)
    return result
```

```python
def dfs_iterative(root: TreeNode, order: str = "preorder") -> list[int]:
    # DFS (обход в глубину) — итеративная версия
    # Идея: явный стек (LIFO) вместо стека вызовов рекурсии
    if not root:
        return []

    if order == "preorder":
        # Корень → левое → правое
        result = []
        stack = [root]

        while stack:
            node = stack.pop()  # берём узел с вершины стека
            result.append(node.val)
            # Сначала кладём правого, потом левого — при pop() левый обработается первым
            if node.right:
                stack.append(node.right)
            if node.left:
                stack.append(node.left)

        return result

    if order == "inorder":
        # Левое → корень → правое
        result = []
        stack = []
        current = root

        while stack or current:
            # Идём влево до конца, сохраняя узлы в стек
            while current:
                stack.append(current)
                current = current.left
            # Обрабатываем узел и переходим в правое поддерево
            current = stack.pop()
            result.append(current.val)
            current = current.right

        return result

    if order == "postorder":
        # Левое → правое → корень
        result = []
        # visited=False — узел ещё не обработан; True — оба ребёнка уже пройдены
        stack = [(root, False)]

        while stack:
            node, visited = stack.pop()
            if visited:
                result.append(node.val)
            else:
                # Откладываем узел «на потом» и кладём детей в стек
                stack.append((node, True))
                if node.right:
                    stack.append((node.right, False))
                if node.left:
                    stack.append((node.left, False))

        return result

    raise ValueError("order must be 'preorder', 'inorder', or 'postorder'")
```
