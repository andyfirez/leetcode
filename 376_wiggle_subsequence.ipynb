Идея:
- Вместо того чтобы хранить все возможные подпоследовательности, мы отслеживаем всего две величины up и down
- up - длина самой длинной wiggle-последовательности, которая заканчивается ростом
- down - длина самой длинной wiggle-последовательности, которая заканчивается падением
- поскольку рост и падение должны чередоваться, то если наблюдаем рост, добавляем к макс. длине падающей, если падение то наоборот
- Здесь используется идея жадного алгоритма: мы храним только длину максимальной последовательности на каждом шаге, увеличивая ее

```python
def wiggleMaxLength(self, nums):
    n = len(nums)
    if n < 2:
        return n
    up = 1 # длина самой длинной wiggle-последовательности, которая заканчивается ростом
    down = 1 # длина самой длинной wiggle-последовательности, которая заканчивается падением
    for i in range(1, n):
        # Мы можем продлить любую последовательность, закончившуюся падением, добавив текущий элемент
        if nums[i] > nums[i - 1]:
            # либо продлеваем ее, либо оставляем текущую длину как максимум
            up = max(up, down + 1)
        # Мы можем продлить любую последовательность, закончившуюся ростом, добавив текущий элемент
        elif nums[i] < nums[i - 1]:
            # либо продлеваем ее, либо оставляем текущую длину как максимум
            down = max(down, up + 1)
            
    return max(up, down)
```

```python
nums = [1]
print(wiggleMaxLength(nums)) #1
```
