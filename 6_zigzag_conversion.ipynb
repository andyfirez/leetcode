Идея: если каждому символу в исходной строке мы поставим в соответствии индекс строки искомой таблицы, то можем заметить, что существует определенная закономерность. индексы растут от меньшего к большему до numRows-1, затем уменьшаются и так происходит цикличто до последнего символа. Алгоритм:
- Для каждого символа строки определить индекс
- положить в словарь индексы в качестве ключей и символы в качестве значений (список)
- вывести символы друг за другом по отсортированным ключам

Сложность: nlogn за счет сортировки

```python
def create_indexes_to_chars(s: str, numRows: int) -> dict:
    is_increasing = True
    i = 0
    indexes_to_chars = {}
    for char in s:
        if i not in indexes_to_chars:
            indexes_to_chars[i] = ""
        indexes_to_chars[i]+=char
        if is_increasing:
            if i >= numRows-1:
                is_increasing = False
        else:
            if i == 0:
                is_increasing = True

        if is_increasing:
            i+=1
        else:
            i-=1

    return indexes_to_chars

def convert(s: str, numRows: int) -> str:
    if numRows==1:
        return s
    indexes_to_chars = create_indexes_to_chars(s, numRows)
    result = ""
    for index, characters in sorted(indexes_to_chars.items()):
        result += characters
    return result

```

Решение 2: Нам нет смысла использовать словарь, мы можем просто складывать символы в отдельные строки и потом обращаться к ним по индексу

```python
def convert(s, numRows):
    if numRows == 1:
        return s
    ans = [""] * numRows
    i, step = 0, None
    for c in s:
        ans[i] += c
        if i == 0:
            step = 1
        elif i == numRows - 1:
            step = -1
        i += step
    return "".join(ans)
```

```python
create_indexes_to_chars("AB", 1)
```

```python
convert("PAYPALISHIRING", 4)
```
