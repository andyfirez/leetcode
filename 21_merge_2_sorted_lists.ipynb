```python
from typing import Optional
```

```python
class ListNode:
    def __init__(self, val=0, next=None):
        self.val = val
        self.next = next
```

```python
def create_linked_list(arr):
    """
    Создает односвязный список из массива.
    
    Args:
        arr: список значений для узлов
    
    Returns:
        ListNode: голову созданного списка или None, если массив пуст
    """
    if not arr:  # проверка на пустой массив
        return None
    
    # Создаем голову списка
    head = ListNode(arr[0])
    current = head
    
    # Проходим по остальным элементам массива
    for i in range(1, len(arr)):
        current.next = ListNode(arr[i])
        current = current.next
    
    return head
```

```python
def linked_list_to_array(head):
    """
    Преобразует односвязный список в массив (список Python).
    
    Args:
        head: голова односвязного списка
    
    Returns:
        list: массив значений узлов списка
    """
    result = []
    current = head
    
    while current:
        result.append(current.val)
        current = current.next
    
    return result
```

```python
def mergeTwoLists(list1: Optional[ListNode], list2: Optional[ListNode]) -> Optional[ListNode]:
    # начальная нода для старта
    dummy = ListNode(-1)
    p = dummy
    while list1 and list2:
        if list1.val <= list2.val:
            p.next = list1 # присоединяем элемент первого списка к результирующему
            list1 = list1.next # смещаем указатель первого списка
            p = p.next # смещаем указатель результирующего списка
        else:
            p.next = list2
            list2 = list2.next
            p = p.next
    if list1:
        p.next = list1
    if list2:
        p.next = list2
    return dummy.next

```

```python
list1 = create_linked_list([1,2,4])
list2 = create_linked_list([1,3,4])
print(linked_list_to_array(mergeTwoLists(list1, list2)))
```
