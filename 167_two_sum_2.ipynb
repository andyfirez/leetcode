```python
from typing import List
```

- поскольку массив отсортирован, при движении вправо наша разница между таргетом и элементом будет гарантированно уменьшаться
- используем 2 указателя и в зависимости от величины разницы сдвигаем либо левый, либо правый указатель

```python
def twoSum(numbers: List[int], target: int) -> List[int]:
    p1=0
    p2=len(numbers)-1
    while p1<=p2:
        d1=target-numbers[p1]
        d2=target-numbers[p2]

        if d1 == numbers[p2] or d2 == numbers[p1]:
            return [p1+1, p2+1]
        
        if d1 < numbers[p2]:
            p2-=1

        if d2 > numbers[p1]:
            p1+=1
    
```

```python
numbers = [2,7,11,15]
target = 9
print(twoSum(numbers, target))
```

```python
numbers = [2,3,4]
target = 6
print(twoSum(numbers, target))
```

```python
numbers = [-1,0]
target = -1
print(twoSum(numbers, target))
```
