```python
from typing import Optional, List
```

```python
class TreeNode:
    def __init__(self, val=0, left=None, right=None):
        self.val = val
        self.left = left
        self.right = right
```

```python
def list_to_tree(tree_list: list) -> Optional[TreeNode]:
    if not tree_list:
        return None

    def traverse(pointer: int) -> Optional[TreeNode]:
       if pointer > len(tree_list)-1:
          return None
       
       left_child = traverse(2*pointer + 1)
       right_child = traverse(2*pointer + 2)
       if not tree_list[pointer]:
           return None 
       return TreeNode(tree_list[pointer], left_child, right_child)

    return traverse(0)
```

```python
def getMinimumDifference(root: Optional[TreeNode]) -> int:
    if not root:
        return 0

    tree_list = []

    def traverse(node: TreeNode):
        if not node:
            return

        traverse(node.left)
        tree_list.append(node.val)
        traverse(node.right)

    traverse(root)

    min_difference = 100001
    for i in range(0, len(tree_list)-1):
        min_difference = min(min_difference, tree_list[i+1] - tree_list[i])

    return min_difference
```

```python
root = [4,2,6,1,3]
print(getMinimumDifference(list_to_tree(root)))
```

```python
root = [1,0,48,None,None,12,49]
print(getMinimumDifference(list_to_tree(root)))
```
