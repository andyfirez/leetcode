```python
# Наивное решение (перебор всех подстрок и проверка на палиндром)
def longestPalindrome(s: str) -> str:
    longest_palindrome = ""
    for i in range(len(s)):
        for j in range(0, len(s)-i):
            substring = s[i:i+j+1]
            reversed_substring = substring[::-1]
            if substring == reversed_substring and len(substring) > len(longest_palindrome):
                longest_palindrome = substring
    return longest_palindrome  
```

```python
longestPalindrome("babad")
```

Алгоритм расширения от центра
Основная идея:
Каждый палиндром имеет центр. Для нечетных палиндромов центр - это один символ, для четных - два одинаковых символа.

Как работает:
- Проходим по каждому символу строки (представляя его как потенциальный центр)
- Для каждого центра пробуем два варианта:
- Нечетная длина: центр = один символ (левая и правая границы равны i)
- Четная длина: центр = два символа (левая граница = i, правая = i+1)
- Расширяемся в обе стороны, пока символы совпадают
- Запоминаем самый длинный найденный палиндром

Временная сложность: O(n²)
Пространственная сложность: O(1)

```python
def longestPalindrome(s: str) -> str:
    if not s:
        return ""
    
    def expand_around_center(left: int, right: int) -> int:
        """Возвращает длину палиндрома с центром в left/right"""
        while left >= 0 and right < len(s) and s[left] == s[right]:
            left -= 1
            right += 1
        return right - left - 1
    
    start = 0
    end = 0
    
    for i in range(len(s)):
        # Палиндром нечетной длины (центр - один символ)
        len1 = expand_around_center(i, i)
        # Палиндром четной длины (центр - два символа)
        len2 = expand_around_center(i, i + 1)
        
        max_len = max(len1, len2)
        
        if max_len > (end - start):
            start = i - (max_len - 1) // 2
            end = i + max_len // 2
    
    return s[start:end + 1]
```
