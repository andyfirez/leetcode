Idea:
- reverse the entire array
- reverse the two parts separately
- use 2 pointers to reverse

```python
def reverse(array, left, right):
    p_left = left
    p_right = right

    while p_left < p_right:
        # temp = array[p_left]
        # array[p_left] = array[p_right]
        # array[p_right] = temp
        array[p_left], array[p_right] = array[p_right], array[p_left]
        p_left+=1
        p_right-=1


def rotate(nums, k):

    k=k%len(nums)

    reverse(nums, 0, len(nums)-1)
    reverse(nums, 0, k-1)
    reverse(nums, k, len(nums)-1)

    return nums
```

```python
nums = [1,2,3,4,5,6,7]
k = 3
print(rotate(nums, k))
```
