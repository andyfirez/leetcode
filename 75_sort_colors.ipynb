```python
from typing import List
```

```python
def sortColors(nums: List[int]) -> None:
    n = len(nums)
    lp = 0
    rp = n-1

    while lp < rp:
        if nums[lp] > nums[rp]:
            temp = nums[rp]
            nums[rp] = nums[lp]
            nums[lp] = temp

            lp+=1
            rp-=1
        else:
            lp+=1

    return nums
```

```python
nums = [2,0,2,1,1,0]
print(sortColors(nums))
```

```python
nums = [2,0,1]
print(sortColors(nums))
```
