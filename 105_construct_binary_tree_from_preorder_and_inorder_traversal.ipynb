```python
from typing import List, Optional
```

```python
class TreeNode:
    def __init__(self, val=0, left=None, right=None):
        self.val = val
        self.left = left
        self.right = right
```

```python
def dfs_recursive(root: TreeNode, order: str = "preorder") -> list[int]:
    # DFS (обход в глубину) — рекурсивная версия
    # Идея: уходим в поддерево до конца, затем возвращаемся (call stack = стек обхода)
    if not root:
        return []

    result = []

    def traverse(node: TreeNode) -> None:
        if not node:
            result.append(None)
            return

        if order == "preorder":
            # Корень → левое → правое
            result.append(node.val)
            traverse(node.left)
            traverse(node.right)
        elif order == "inorder":
            # Левое → корень → правое
            traverse(node.left)
            result.append(node.val)
            traverse(node.right)
        elif order == "postorder":
            # Левое → правое → корень
            traverse(node.left)
            traverse(node.right)
            result.append(node.val)
        else:
            raise ValueError("order must be 'preorder', 'inorder', or 'postorder'")

    traverse(root)
    return result
```

```python
def buildTree(preorder: List[int], inorder: List[int]) -> Optional[TreeNode]:
    if not preorder:
        return None

    value_to_inorder_index = {}
    for i in range(len(inorder)):
        value_to_inorder_index[inorder[i]] = i

    def traverse(preorder_left: int, preorder_right: int, inorder_left: int, inorder_right: int) -> Optional[TreeNode]:
        if preorder_left > preorder_right:
            return None
            
        inorder_index = value_to_inorder_index[preorder[preorder_left]]
        
        left_subtree_size = inorder_index - inorder_left
        
        left_child = traverse(
            preorder_left + 1, 
            preorder_left + left_subtree_size,
            inorder_left,
            inorder_index - 1
        )
        
        right_child = traverse(
            preorder_left + left_subtree_size + 1, 
            preorder_right,
            inorder_index + 1,
            inorder_right
        )
        
        return TreeNode(preorder[preorder_left], left_child, right_child)

    return traverse(0, len(preorder) - 1, 0, len(inorder) - 1)
    
```

```python
preorder = [3,9,20,15,7]
inorder = [9,3,15,20,7]
print(dfs_recursive(buildTree(preorder, inorder)))
```

```python
preorder = [-1]
inorder = [-1]
print(dfs_recursive(buildTree(preorder, inorder)))
```
