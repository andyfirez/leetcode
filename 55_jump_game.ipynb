```python
def canJump(nums):
    pointer = 0
    while pointer <= len(nums):
        if nums[pointer]==0:
            return False
        max_jump=0
        for i in range(nums[pointer]):

            if pointer+i < len(nums):
                if nums[pointer+i] > max_jump:
                    max_jump = nums[pointer+i]

        pointer+=max_jump

    return True
```

```python
nums = [2,3,1,1,4]
print(canJump(nums))
```

```python
nums = [3,2,1,0,4]
print(canJump(nums))
```
