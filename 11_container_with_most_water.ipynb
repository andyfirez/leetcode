```python
from typing import List
```

Идея:
- Начинаем перебор с самого широкого контейнера и ставим указатели на левую и правую границы
- Имеет смысл двигать только меньшую по высоте границу, поскольку она задает максимум возможной воды. Если будем двигать большую, то даже еслти найдем более высокую, то кол-во воды не увеличится
- Двигаем 2 указателя навстречу друг другу, пока они не встретятся. Выводим максимум площади

```python
def maxArea(height: List[int]) -> int:
    result = 0
    left_pointer = 0
    right_pointer = len(height)-1
    while left_pointer < right_pointer:
        if height[left_pointer] < height[right_pointer]:
            result = max(result, (right_pointer-left_pointer)*height[left_pointer])
            left_pointer+=1
        else:
            result = max(result, (right_pointer-left_pointer)*height[right_pointer])
            right_pointer-=1

    return result

```

```python
height = [1,8,6,2,5,4,8,3,7]
max_area = maxArea(height)
max_area
```
