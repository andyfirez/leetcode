## Cубоптимальное решение
Сложность - оптимизированный квадрат
Алгоритм:
1. Пробегаем по строке до первого дубликата, сохраняем в словарь интекс каждого символа
2. Как только нашли дубликат - сохраняем значение и "сдвигаем" наш индекс перебора к первому вхождению дубликата. До него нет смысла перебирать символы. Так мы оптимизируем квадратичный перебор.
3. Возвращаем максимальную длину

```python
def lengthOfLongestSubstring(s: str) -> int:
    if not s:
        return 0
    char_dict = {} 
    counter = 0
    max_count = 0
    for i, char in enumerate(s):
        if char not in char_dict:
            char_dict[char] = i
            counter+=1
        else:
            if max_count < counter:
                max_count = counter
            duplicate_index = char_dict[char]
            char_dict = {}
            for j, char in enumerate(s[duplicate_index:i+1]):
                char_dict[char] = duplicate_index+j
            counter = len(char_dict)

    if max_count < counter:
        return counter

    return max_count

lengthOfLongestSubstring("bpfbhmipx")
```

```python
def lengthOfLongestSubstring(s: str) -> int:
    ans = 0
    left_bound = 0
    char_dict = {}
    for i, char in enumerate(s):
        # Если символ не является дубликатом или за пределами текущего окна
        if char not in char_dict or char_dict[char] < left_bound:
            # Увеличиваем счетчик на 1
            ans = max(ans, i - left_bound + 1)
        else:
            # если символ является дубликатом, сдвигаем левую границу окна к текущему символу
            left_bound = char_dict[char] + 1

        # Запоминаем индекс символа
        char_dict[char] = i
    return ans
            

lengthOfLongestSubstring("abcabcbb")
```
