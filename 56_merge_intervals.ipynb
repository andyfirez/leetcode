```python
from typing import List
```

Идея:
- Сортируем отрезки в порядке возрастания начала. Таким образом мы гарантируем, что каждый следущий отрезок начинается не раньше текущего
- для текущего отрезка проверяем, пересекается ли он со следующим. если да - сливаем их в один и продолжаем
- как только обнаружили отсутствие пересечения, сохраняем результат в массив

```python
def merge(intervals: List[List[int]]) -> List[List[int]]:
    intervals.sort(key=lambda interval: interval[0])

    result = []
    start_merge = intervals[0][0]
    end_merge = intervals[0][1]
    for i in range(1, len(intervals)):
        interval = intervals[i]
        if interval[1] < end_merge:
            continue
        elif start_merge <= interval[0] <= end_merge or interval[0] <= end_merge <= interval[1]:
            end_merge = interval[1]
        else:
            result.append([start_merge, end_merge])
            start_merge = interval[0]
            end_merge=interval[1]

    result.append([start_merge, end_merge])
    
    return result

```

```python
intervals = [[1,3],[2,6],[8,10],[15,18]]
print(merge(intervals))
```

Упрощенная реализация без избыточных проверок:
- start_merge <= interval[0] - лишняя, потому что сортировка и так это гарантирует

```python
def merge(self, intervals):
    intervals.sort(key=lambda x: x[0])
    ans = []
    for i in intervals:
        if ans and i[0] <= ans[-1][1]:
            ans[-1][1] = max(ans[-1][1], i[1])
        else:
            ans.append(i)
    return ans
```
