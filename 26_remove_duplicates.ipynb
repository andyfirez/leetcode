```python
from typing import List
```

```python
# наивное решение
def removeDuplicates(nums: List[int]) -> int:
    if len(nums) == 1:
        return 1, nums
    i=0
    j=0
    while i< len(nums)-1 and j<len(nums):
        j=i+1
        while nums[j] <= nums[i]:
            if j==len(nums)-1:
                return i+1, nums
            j+=1
        nums[i+1] = nums[j]
        i+=1

        if j==len(nums)-1:
            return i+1, nums
    return i, nums
```

Идея:
- пробегаем все элементы один раз, сравнивая с последним вставленным
- последний вставленный элемент является критерием для сравнения, поскольку следующий вставленный обязан быть больше него
- Храним индекс места для вставки и сдвигаем его при вставке

```python
def removeDuplicates(self, nums):
    idx = 0 # позиция для вставки следующего уникального элемента
    for num in nums:
        if idx < 1 or num != nums[idx - 1]: # проверяем, нужно ли вставлять
            nums[idx] = num # вставляем элемент на позицию idx
            idx += 1 # перемещаем указатель вставки дальше
    return idx
```
