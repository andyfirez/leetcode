Идея:
- Решается по аналогии с 438 задачей про анаграммы
- также используются 2 словаря со счетчиками. Один для меньшей строки. Второй для скользящего окна

```python
def checkInclusion(s1: str, s2: str) -> bool:
    if len(s1) > len(s2):
        return False
    s1_counts = {}
    window_counts = {}
    for c in s1:
        s1_counts[c] = s1_counts.get(c, 0) + 1

    for i in range(len(s2)):
        window_counts[s2[i]] = window_counts.get(s2[i], 0) + 1

        if i>= len(s1):
            left_char = s2[i - len(s1)]
            if window_counts[left_char] == 1:
                del window_counts[left_char]
            else:
                window_counts[left_char] -= 1

        if i>= len(s1)-1 and window_counts == s1_counts:
            return True
        
    return False
        
    
```

```python
s1 = "ab"
s2 = "eidbaooo"
print(checkInclusion(s1, s2)) #True
```
