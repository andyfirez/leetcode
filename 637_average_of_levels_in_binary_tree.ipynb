```python
from typing import Optional, List
```

```python
class TreeNode:
    def __init__(self, val=0, left=None, right=None):
        self.val = val
        self.left = left
        self.right = right
```

```python
def list_to_tree(tree_list: list) -> Optional[TreeNode]:
    if not tree_list:
        return None

    def traverse(pointer: int) -> Optional[TreeNode]:
       if pointer > len(tree_list)-1:
          return None
       
       left_child = traverse(2*pointer + 1)
       right_child = traverse(2*pointer + 2)
       if not tree_list[pointer]:
           return None 
       return TreeNode(tree_list[pointer], left_child, right_child)

    return traverse(0)
```

```python
def averageOfLevels(root: Optional[TreeNode]) -> List[float]:
    if not root:
        return []

    tree: list[list[int]] = []

    def traverse(node: TreeNode, level: int):
        if not node:
            return
        
        if level == len(tree):
            tree.append([])

        tree[level].append(node.val)
        traverse(node.left, level+1)
        traverse(node.right, level+1)

    traverse(root, 0)

    result = [sum(level) / len(level) for level in tree]
    return result

    
```

```python
tree_array = [3,9,20,None,None,15,7]
print(averageOfLevels(list_to_tree(tree_array)))
```
