```python
from typing import List
```

Идея:
- поскольку мы не шортим, то прибыль мы можем получить только за счет дней роста. Максимальная прибыль - это когда мы захватили все дни роста и 0 дней падения
- Это значит, что нам достаточно просто сложить весь прирост за все дни роста и это будет ответ

```python
def maxProfit(prices: List[int]) -> int:
    max_profit = 0
    for i in range(len(prices)-1):
        profit = prices[i+1] - prices[i]
        if profit >0:
            max_profit+=profit

    return max_profit
```

```python
prices = [7,1,5,3,6,4]
print(maxProfit(prices))
```

```python
prices = [1,2,3,4,5]
print(maxProfit(prices))
```

```python
prices = [7,6,4,3,1]
print(maxProfit(prices))
```
