Идея:
- Задачу можно свести к задаче о сумме двух. Мы фиксируем одно число и требуем, чтобы сумма двух других равнялась ему
- Для начала сортируем массив. Потом фиксируем самое маленькое (левое) число. Далее работаем с 2 указателями по оставшейся части и двигаем их по направлению друг к другу, пока они не сойдутся, собирая подходящие пары чисел
- чтобы предотвратить повторы, мы сдвигаем первое число и указатели, если одинаковые числа идут подряд в отсортированном массиве

```python
def threeSum(nums: list[int]) -> list[list[int]]:
    nums.sort()
    n = len(nums)
    result = []
    for i in range(n-2):
        if i > 0 and nums[i] == nums[i - 1]:
            continue
        left_pointer = i+1
        right_pointer = n-1
        target_sum = -nums[i]
        while left_pointer < right_pointer:
            if target_sum == (nums[left_pointer] + nums[right_pointer]):
                result.append([nums[i], nums[left_pointer], nums[right_pointer]])
                left_pointer+=1
                right_pointer-=1
                while nums[left_pointer] == nums[left_pointer-1] and left_pointer < right_pointer:
                    left_pointer+=1
                while nums[right_pointer] == nums[right_pointer+1] and left_pointer < right_pointer:
                    right_pointer-=1
            elif target_sum > nums[left_pointer] + nums[right_pointer]:
                left_pointer+=1
            else:
                right_pointer-=1

    return result
```

```python
nums = [-1,0,1,2,-1,-4]
# nums = [-4,-1,-1,0,1,2]
print(threeSum(nums))
```

```python
nums = [0,1,1]
print(threeSum(nums))
```

```python
nums = [0,0,0]
print(threeSum(nums))
```
