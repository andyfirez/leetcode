Идея:
- строим двухмерную таблицу dp, где dp[i][j] - это ответ для первых i символов word1 и первых j символов word2
- Когда i = 0, это значит, что мы взяли 0 первых символов из word1, то есть пустую строку
- Когда j = 0, это значит, что мы взяли 0 первых символов из word2, то есть пустую строку
- dp[0][j] = j поскольку у нас есть пустая строка (""), и мы хотим превратить её в строку длины j (первые j символов word2)
- dp[i][0] = i поскольку у нас есть строка длины i (первые i символов word1), и мы хотим превратить её в пустую строку ("")
- Замена = 1 + dp[i-1][j-1]
- Удаление = 1 + dp[i-1][j]
- Вставка = 1 + dp[i][j-1]
- Берем минимум из 3 вариантов
- Наша задача - заполнить таблицу, и тогда ответ окажется в ячейке dp[len(word1)][len(word2)]

```python
def minDistance(word1: str, word2: str) -> int:
    rows = len(word1)
    cols = len(word2)

    if rows == cols == 0:
        return 0
    
    if rows == 0:
        return cols
    
    if cols == 0:
        return rows
    
    dp = [[0] * (cols+1) for _ in range(rows+1)]

    for i in range(rows+1):
        dp[i][0] = i

    for j in range(cols+1):
        dp[0][j] = j

    for i in range(1, rows+1):
        for j in range(1, cols+1):
            if word1[i-1] == word2[j-1]:
                dp[i][j] = dp[i-1][j-1]
            else:
                dp[i][j] = min(dp[i-1][j-1], dp[i-1][j], dp[i][j-1]) + 1

    return dp[rows][cols]

        
```

```python
word1 = "horse"
word2 = "ros"
print(minDistance(word1, word2))
```
