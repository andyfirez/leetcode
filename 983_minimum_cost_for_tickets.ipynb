```python
from typing import List
```

```python
def mincostTickets(days: List[int], costs: List[int]) -> int:
    dp = [0] * len(days)
    dp[len(days)] = min(costs[0], costs[1], costs[2])
    for i in range(len(days)-2, -1, -1):
        j=i+1
        while days[j] - days[i] <=7 and j<len(days)-1:
            j+=1

        k=i+1
        while days[k] - days[i] <=30 and k<len(days)-1:
            k+=1

        option_1 = costs[0] + dp[i+1]
        option_7 = costs[1] + dp[j]
        option_30 = costs[2] + dp[k]

        dp[i] = min(option_1, option_7, option_30)
    
    return dp[0]
```

```python
days = [1,4,6,7,8,20]
costs = [2,7,15]
print(mincostTickets(days, costs))
```
