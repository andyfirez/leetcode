```python
from typing import List
```

Идея:
- создаем словарь со счетчиками символов для строки p
- используем скользящее окно, которое реализовано как словарь со счетчиками для каждого символа, который присутствует в диапазоне, равном размеру окна
- пошагово сдвигаем окно по одному символу и на каждой итерации сравниваем со словарем для p. Если они одинаковые, то мы нашли анаграмму
- Чтобы удалить элемент из окна, используем оператор del, это позволяет нам избежать ситуации, когда счетчик равен, нулю, но при этом словари получаются разные

```python
def findAnagrams(s: str, p: str) -> List[int]:
    result = []
    len_s, len_p = len(s), len(p)
    
    if len_s < len_p:
        return result
    
    p_count = {} # частоты символов в строке p
    window_count = {} # скользящее окно с частотами символов
    
    # Заполняем словарь для p
    for char in p:
        p_count[char] = p_count.get(char, 0) + 1
    
    for i in range(len_s):
        # Увеличиваем счетчик или добавляем новый символ в окно
        window_count[s[i]] = window_count.get(s[i], 0) + 1
        
        # Удаляем символ, который вышел из окна
        if i >= len_p:
            left_char = s[i - len_p]
            if window_count[left_char] == 1:
                del window_count[left_char]  # Удаляем ключ, если значение станет 0
            else:
                window_count[left_char] -= 1
        
        # Сравниваем наше скользящее окно со словарем строки, если они равны, то мы нашли анаграмму
        if i >= len_p - 1 and window_count == p_count:
            result.append(i - len_p + 1)
    
    return result
```

```python
s = "cbaebabacd"
p = "abc"
print(findAnagrams(s, p))
```

```python
s = "abab"
p = "ab"
print(findAnagrams(s, p))
```
